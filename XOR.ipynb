```python
import numpy as np
```

```python
X = np.array([[0,0], [1,0], [0,1], [1,1]]).T
y = np.array([[1, 0, 0, 1]])
```

```python
np.random.seed(42)

w1 = np.random.randn(4,2)*0.1
b1 = np.zeros((4,1))

w2 = np.random.randn(4,4)*0.1
b2 = np.zeros((4,1))

w3 = np.random.randn(4,4)*0.1
b3 = np.zeros((4,1))

w4 = np.random.randn(1,4)*0.1
b4 = np.zeros((1,1))

m=X.shape[1]
lr=0.1
```

```python
def sigmoid(z):
    return 1/(1+np.exp(-z))
def relu(z):
    return np.maximum(0,z)
def relu_derivative(z):
    return (z>0).astype(float)
```

```python
for i in range(10000):
    m=X.shape[1]
    #Forword Prop
    Z1=np.dot(w1,X)+b1
    A1=relu(Z1)
    Z2=np.dot(w2,A1)+b2
    A2=relu(Z2)
    Z3=np.dot(w3,A2)+b3
    A3=relu(Z3)
    Z4=np.dot(w4,A3)+b4
    A4=sigmoid(Z4)
    
    #Backword Prop
    dZ4=A4-y
    dw4=(1/m)*np.dot(dZ4,A3.T)
    db4=(1/m)*np.sum(dZ4,axis=1,keepdims=True)

    dA3=np.dot(w4.T,dZ4)
    dZ3=dA3*relu_derivative(Z3)
    dw3=(1/m)*np.dot(dZ3,A2.T)
    db3=(1/m)*np.sum(dZ3,axis=1,keepdims=True)

    dA2=np.dot(w3.T,dZ3)
    dZ2=dA2*relu_derivative(Z2)
    dw2=(1/m)*np.dot(dZ2,A1.T)
    db2=(1/m)*np.sum(dZ2,axis=1,keepdims=True)

    dA1 = np.dot(w2.T, dZ2)
    dZ1 = dA1 * (Z1 > 0)
    dw1 = (1/m) * np.dot(dZ1, X.T)
    db1 = (1/m) * np.sum(dZ1, axis=1, keepdims=True)

    #Weight and bias updateing

    w1-=lr*dw1
    b1-=lr*db1
    w2-=lr*dw2
    b2-=lr*db2
    w3-=lr*dw3
    b3-=lr*db3
    w4-=lr*dw4
    b4-=lr*db4
    if i % 1000 == 0:
        cost = -np.mean(y * np.log(A4 + 1e-8) + (1-y) * np.log(1-A4 + 1e-8))
        print(f"Iteration {i}, Cost: {cost:.4f}")
print("Final Predictions:")
print((A4 > 0.5).astype(int))
```
